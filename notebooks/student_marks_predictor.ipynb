```python
# Create copies and add subject labels

math_data = df_mat.drop(columns=["result"], errors="ignore").copy()
port_data = df_por.drop(columns=["result"], errors="ignore").copy()

math_data["subject"] = "Mathematics"
port_data["subject"] = "Portuguese"

# Combine both datasets
df = pd.concat(
    [math_data, port_data],
    ignore_index=True
)

print("Combined dataset shape:", df.shape)
display(df.head())
```

```python
print("Total records:", len(df))
print("Total columns:", len(df.columns))

print("\nSubject distribution:")
print(df["subject"].value_counts())

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
# Target variable
y = df["G3"]

# Input features
X = df.drop(columns=["G3"])

print("Features shape:", X.shape)
print("Target shape:", y.shape)

display(X.head())
display(y.head())
```

```python
# Identify categorical columns
categorical_features = X.select_dtypes(
    include=["object"]
).columns.tolist()

# Identify numerical columns
numerical_features = X.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

print("Categorical features:")
print(categorical_features)

print("\nNumerical features:")
print(numerical_features)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("Training features:", X_train.shape)
print("Testing features:", X_test.shape)

print("Training target:", y_train.shape)
print("Testing target:", y_test.shape)
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer

# Numerical preprocessing
numerical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Combine both transformations
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessing pipeline created successfully!")
```

```python
from sklearn.linear_model import LinearRegression

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

print("Linear Regression imported successfully!")
```

```python
# Create Linear Regression model
linear_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", LinearRegression())
])

print("Linear Regression pipeline created!")
```

```python
# Train the model
linear_model.fit(X_train, y_train)

print("Linear Regression model trained successfully!")
```

```python
# Predict final grades
y_pred = linear_model.predict(X_test)

print("Actual grades:", y_test.values[:10])
print("Predicted grades:", np.round(y_pred[:10], 2))
```

```python
# Calculate evaluation metrics

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("Linear Regression Performance")
print("-----------------------------")
print("MAE:", round(mae, 2))
print("MSE:", round(mse, 2))
print("RMSE:", round(rmse, 2))
print("R2 Score:", round(r2, 4))
```

```python
plt.figure(figsize=(8, 6))

plt.scatter(y_test, y_pred, alpha=0.6)

plt.plot(
    [0, 20],
    [0, 20],
    linestyle="--"
)

plt.xlabel("Actual Grades")
plt.ylabel("Predicted Grades")
plt.title("Actual vs Predicted Student Grades")

plt.show()
```

```python
from sklearn.ensemble import RandomForestRegressor

print("Random Forest imported successfully!")
```

```python
rf_model = Pipeline(steps=[
    ("preprocessor", preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=200,
        max_depth=10,
        min_samples_split=5,
        random_state=42,
        n_jobs=-1
    ))
])

print("Random Forest pipeline created!")
```

```python
rf_model.fit(X_train, y_train)

print("Random Forest trained successfully!")
```

```python
rf_pred = rf_model.predict(X_test)

print("Actual grades:", y_test.values[:10])
print("Predicted grades:", np.round(rf_pred[:10], 2))
```

```python
rf_mae = mean_absolute_error(y_test, rf_pred)

rf_mse = mean_squared_error(y_test, rf_pred)

rf_rmse = np.sqrt(rf_mse)

rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Performance")
print("-------------------------")
print("MAE:", round(rf_mae, 2))
print("MSE:", round(rf_mse, 2))
print("RMSE:", round(rf_rmse, 2))
print("R2 Score:", round(rf_r2, 4))
```

```python
model_comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Random Forest"],
    "MAE": [mae, rf_mae],
    "RMSE": [rmse, rf_rmse],
    "R2 Score": [r2, rf_r2]
})

display(model_comparison.round(4))
```

```python
# Remove G1 and G2 to prevent using previous exam grades

X_early = df.drop(columns=["G1", "G2", "G3"])

y_early = df["G3"]

print("Early-warning features:", X_early.shape)
print("Target:", y_early.shape)

print("G1 included:", "G1" in X_early.columns)
print("G2 included:", "G2" in X_early.columns)
```

```python
X_train_early, X_test_early, y_train_early, y_test_early = train_test_split(
    X_early,
    y_early,
    test_size=0.20,
    random_state=42
)

print("Training data:", X_train_early.shape)
print("Testing data:", X_test_early.shape)
```

```python
early_num_features = X_early.select_dtypes(
    include=np.number
).columns.tolist()

early_cat_features = X_early.select_dtypes(
    include=["object"]
).columns.tolist()

early_preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, early_num_features),
        ("cat", categorical_transformer, early_cat_features)
    ]
)

print("Early-warning preprocessing ready!")
```

```python
early_model = Pipeline(steps=[
    ("preprocessor", early_preprocessor),
    ("model", RandomForestRegressor(
        n_estimators=200,
        max_depth=10,
        random_state=42,
        n_jobs=-1
    ))
])

early_model.fit(X_train_early, y_train_early)

print("Early-warning model trained successfully!")
```

```python
early_pred = early_model.predict(X_test_early)

early_mae = mean_absolute_error(y_test_early, early_pred)
early_rmse = np.sqrt(mean_squared_error(y_test_early, early_pred))
early_r2 = r2_score(y_test_early, early_pred)

print("EARLY WARNING MODEL")
print("-------------------")
print("MAE:", round(early_mae, 2))
print("RMSE:", round(early_rmse, 2))
print("R2 Score:", round(early_r2, 4))
```

```python
def detect_weak_areas(student):

    weak_areas = []

    if student["studytime"] <= 2:
        weak_areas.append("Needs study-time improvement")

    if student["absences"] > 10:
        weak_areas.append("High number of absences")

    if student["failures"] > 0:
        weak_areas.append("Previous academic failures")

    if student["G1"] < 10:
        weak_areas.append("Low first-period grade")

    if student["G2"] < 10:
        weak_areas.append("Low second-period grade")

    if len(weak_areas) == 0:
        weak_areas.append("No major risk indicators detected")

    return weak_areas


print("Weak-area detection function created!")
```

```python
student = df_mat.iloc[0]

weak_areas = detect_weak_areas(student)

print("Student academic indicators:")

for area in weak_areas:
    print("-", area)
```

```python
# 1 = Pass, 0 = Fail

y_pass = (df["G3"] >= 10).astype(int)

# Use early-warning features (without G1 and G2)
X_class = X_early.copy()

print("Pass/Fail target created!")

print("\nClass distribution:")
print(y_pass.value_counts())

print("\nPass percentage:")
print(round(y_pass.mean() * 100, 2), "%")
```

```python
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class,
    y_pass,
    test_size=0.20,
    random_state=42,
    stratify=y_pass
)

print("Training data:", X_train_class.shape)
print("Testing data:", X_test_class.shape)

print("\nTraining class distribution:")
print(y_train_class.value_counts())
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(steps=[
    ("preprocessor", early_preprocessor),
    ("model", LogisticRegression(max_iter=2000))
])

logistic_model.fit(X_train_class, y_train_class)

print("Logistic Regression trained successfully!")
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_classifier = Pipeline(steps=[
    ("preprocessor", early_preprocessor),
    ("model", RandomForestClassifier(
        n_estimators=200,
        max_depth=10,
        min_samples_split=5,
        random_state=42,
        n_jobs=-1
    ))
])

rf_classifier.fit(X_train_class, y_train_class)

print("Random Forest Classifier trained successfully!")
```

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report
)

# Predictions
logistic_pred = logistic_model.predict(X_test_class)
rf_class_pred = rf_classifier.predict(X_test_class)

def evaluate_classifier(name, actual, predicted):

    print("\n", name)
    print("-" * 35)

    print("Accuracy:",
          round(accuracy_score(actual, predicted), 4))

    print("Precision:",
          round(precision_score(actual, predicted, zero_division=0), 4))

    print("Recall:",
          round(recall_score(actual, predicted, zero_division=0), 4))

    print("F1 Score:",
          round(f1_score(actual, predicted, zero_division=0), 4))


evaluate_classifier(
    "Logistic Regression",
    y_test_class,
    logistic_pred
)

evaluate_classifier(
    "Random Forest Classifier",
    y_test_class,
    rf_class_pred
)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

ConfusionMatrixDisplay.from_predictions(
    y_test_class,
    logistic_pred,
    display_labels=["Fail", "Pass"],
    ax=axes[0],
    cmap="Blues"
)

axes[0].set_title("Logistic Regression")

ConfusionMatrixDisplay.from_predictions(
    y_test_class,
    rf_class_pred,
    display_labels=["Fail", "Pass"],
    ax=axes[1],
    cmap="Greens"
)

axes[1].set_title("Random Forest Classifier")

plt.tight_layout()
plt.show()
```

```python
classification_comparison = pd.DataFrame({
    "Model": [
        "Logistic Regression",
        "Random Forest Classifier"
    ],
    "Accuracy": [
        accuracy_score(y_test_class, logistic_pred),
        accuracy_score(y_test_class, rf_class_pred)
    ],
    "Precision": [
        precision_score(y_test_class, logistic_pred, zero_division=0),
        precision_score(y_test_class, rf_class_pred, zero_division=0)
    ],
    "Recall": [
        recall_score(y_test_class, logistic_pred, zero_division=0),
        recall_score(y_test_class, rf_class_pred, zero_division=0)
    ],
    "F1 Score": [
        f1_score(y_test_class, logistic_pred, zero_division=0),
        f1_score(y_test_class, rf_class_pred, zero_division=0)
    ]
})

display(classification_comparison.round(4))
```

```python
import joblib
import pandas as pd
from pathlib import Path

ARTIFACT_DIR = Path.cwd().parent / "artifacts"
ARTIFACT_DIR.mkdir(parents=True, exist_ok=True)

# Save the exact input columns used by the early-warning models
joblib.dump(
    X_early.columns.tolist(),
    ARTIFACT_DIR / "early_warning_features.pkl"
)

# Create a baseline profile using median/mode values
default_profile = {}

for col in X_early.columns:
    if pd.api.types.is_numeric_dtype(X_early[col]):
        default_profile[col] = float(X_early[col].median())
    else:
        default_profile[col] = X_early[col].mode(dropna=True).iloc[0]

joblib.dump(
    default_profile,
    ARTIFACT_DIR / "default_student_profile.pkl"
)

print("Feature information saved successfully!")
print("Number of input features:", len(X_early.columns))
print("Artifact folder:", ARTIFACT_DIR.resolve())
```
